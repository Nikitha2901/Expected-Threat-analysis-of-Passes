```python
import pandas as pd
```

#### 2023 data

```python
event_2023 = pd.read_csv("event_data_2023.csv")
```

```python
event_2023
```

```python
event_features_2023 = pd.read_csv("event_features_2023.csv")
```

```python
event_features_2023["actual_xG"] = event_2023["possession_xg"]
```

```python
event_features_2023['possession_goal'] = event_2023['possession_goal']
```

```python
event_features_2023 = event_features_2023.rename(columns={"passer distance goal": "start_distance_to_goal",
    "pass end distance to goal" : "end_distance_to_goal",
    "pass length": "pass_length",
    "pass angle": "pass_angle",
    "passer angle goal": "start_angle_to_goal",
    "pass end angle to goal": "end_angle_to_goal",
    "Passer distance to sideline": "start_distance_to_sideline",
    "pass end distance to sideline": "end_distance_to_sideline", 
    "opponents behind passer": "opponents_behind",
    "teammates behind passer": "teammates_behind",
    "total pressure passer": "pressure_on_passer",
    "teammates beyond passer": "teammates_beyond",
    "opponents beyond passer": "opponents_beyond",
    "opponents between": "Opponents_between",
    "opponents bypass new": "packing",
    "opponents near the passer": "opponents_nearby",
    "average speed team": "average_speed_of_teammates",
    "average speed opponent": "average_speed_of_opponents",
    "teammates near passer": "teammates_nearby"})
```

```python
event_features_2023.columns
```

```python
tracking_2023 = pd.read_csv("tracking_data_2023.csv")
```

```python
tracking_2023
```

```python
tracking_2023.columns
```

```python
#Get passer name mapping
passer_rows = tracking_2023.merge(event_features_2023, on=['id', 'player_id'], how='inner')
passer_names = passer_rows[['id', 'player_id', 'player_name']].drop_duplicates()

#Merge passer name
event_with_passer = event_features_2023.merge(
    passer_names,
    on=['id', 'player_id'],
    how='left'
).rename(columns={'player_name': 'passer_name'})

# Get receiver name mapping
receiver_lookup = tracking_2023.merge(
    event_features_2023[['id', 'pass_recipient_id']].drop_duplicates(),
    left_on=['id', 'player_id'],
    right_on=['id', 'pass_recipient_id'],
    how='inner'
)
receiver_names = receiver_lookup[['id', 'player_id', 'player_name']].drop_duplicates()

# Merge receiver name into the result that already has passer_name
event_features_2023_final = event_with_passer.merge(
    receiver_names,
    left_on=['id', 'pass_recipient_id'],
    right_on=['id', 'player_id'],
    how='left'
).rename(columns={'player_name': 'receiver_name'}).drop(columns=['player_id'], errors='ignore')
```

```python
event_features_2023_final.columns
```

```python
event_features_2023_final.drop(['player_id_y'],axis=1,inplace=True)
```

```python
event_features_2023_final = event_features_2023_final.rename(columns={'player_id_x' : 'player_id'})
```

```python
event_features_2023_final.columns
```

```python
event_features_2023_final
```

```python
team_2023 = pd.read_csv("team_ids_2023.csv")
```

```python
team_2023
```

```python
event_features_2023_final = event_features_2023_final.merge(
    team_2023[['team_id', 'team_name']],
    on='team_id',
    how='left'  # keeps all events even if no matching team found
)
```

```python
event_features_2023_final = event_features_2023_final[['id', 'match_id', 'team_id', 'possession_team_id', 'player_id', 'passer_name', 'pass_recipient_id',
       'receiver_name', 'team_name','start_x', 'start_y', 'end_x', 'end_y', 'passer_x', 'passer_y',
       'start_distance_to_goal','end_distance_to_goal', 'pass_length', 'pass_angle','start_angle_to_goal', 'end_angle_to_goal','start_distance_to_sideline', 'end_distance_to_sideline',
       'forward pass', 'backward pass', 'lateral pass', 'teammates_behind','teammates_beyond', 'opponents_beyond', 'opponents_behind',
       'opponents_between', 'packing', 'pressure_on_passer','average_speed_of_teammates', 'average_speed_of_opponents','opponents_nearby','pressure level passer',
       'teammates_nearby', 'actual_xG', 'possession_goal','possession_xG_target'
       ]]
```
